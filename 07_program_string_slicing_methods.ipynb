# 📝 **Program No. 07 – Demonstrate String Slicing and String Methods**

## 🟢 **Creating a String:**
**A sample string is stored in a variable and used to demonstrate common Python string operations.**

## 🔵 **Using Indexing and Slicing:**
**String indexing accesses individual characters, while slicing extracts a selected range of characters.**

## 🟡 **Changing String Case:**
**Methods such as `upper()`, `lower()`, `title()`, and `swapcase()` demonstrate different ways to change character case.**

## 🟠 **Searching and Replacing Text:**
**Methods such as `find()`, `count()`, `replace()`, `startswith()`, and `endswith()` are used to search and modify string content.**

## 🔴 **Splitting and Character Checks:**
**`split()` and `join()` manipulate text, while methods such as `isalpha()`, `isdigit()`, and `islower()` check string properties.**


```python
text = "Python Programming Language"

print("Original:", text)
print("Length:", len(text))
print("First character:", text[0])
print("First 6 characters:", text[:6])
print("Last 8 characters:", text[-8:])
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title:", text.title())
print("Swapcase:", text.swapcase())
print("Find 'Programming':", text.find("Programming"))
print("'Python' in text:", "Python" in text)
print("Starts with 'Python':", text.startswith("Python"))
print("Ends with 'Language':", text.endswith("Language"))
print("Replace:", text.replace("Python", "Java"))
print("Split:", text.split())
print("Join:", "-".join(text.split()))
print("Count of 'a':", text.lower().count("a"))
print("Is alphabetic:", text.isalpha())
print("Is digit:", text.isdigit())
print("Is lowercase:", text.islower())
```
