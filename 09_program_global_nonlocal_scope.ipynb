# 📝 **Program No. 09 – Demonstrate Global and Nonlocal Variable Scope**

## 🟢 **Global Variable:**
**The variable `x` is created outside all functions, making it a global variable that can be accessed throughout the program.**

## 🔵 **Outer Function Variable:**
**The variable `y` is created inside `outer_function()`, so it belongs to the outer function's local scope.**

## 🟡 **Using the `global` Keyword:**
**The `global` keyword allows the inner function to modify the global variable `x`.**

## 🟠 **Using the `nonlocal` Keyword:**
**The `nonlocal` keyword allows the inner function to modify `y`, which belongs to the enclosing `outer_function()`.**

## 🔴 **Displaying Final Values:**
**The program prints the values inside the inner and outer functions and finally displays the modified global value outside the functions.**


```python
x = 10

def outer_function():
    y = 20

    def inner_function():
        global x
        nonlocal y

        x = 100
        y = 200

        print("Inside inner function:")
        print("Global x:", x)
        print("Nonlocal y:", y)

    inner_function()

    print("\nInside outer function:")
    print("Nonlocal y:", y)

outer_function()

print("\nOutside functions:")
print("Global x:", x)
```
