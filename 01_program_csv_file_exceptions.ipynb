# 📝 **Program No. 01 – Create a CSV File, Read It, and Handle File Exceptions**

## 🟢 **Creating the CSV File:**
**The `csv` module is used to create an employee CSV file and store records in rows with appropriate column headings.**

## 🔵 **Writing Records:**
**`csv.DictWriter` writes employee details into the CSV file and `writeheader()` adds the field names as the first row.**

## 🟡 **Reading the CSV File:**
**`csv.DictReader` reads the stored records and displays each employee record in a readable format.**

## 🟠 **Handling File Exceptions:**
**The program handles `FileNotFoundError`, `PermissionError`, `OSError`, and CSV-related errors so that file problems do not terminate the program unexpectedly.**

## 🔴 **Displaying the Results:**
**After writing and reading the file, the program displays the stored employee information and confirms successful file processing.**


```python
import csv

FILE_NAME = "employees.csv"

employees = [
    {"employee_id": 101, "name": "Arun", "department": "IT", "salary": 45000},
    {"employee_id": 102, "name": "Meena", "department": "HR", "salary": 42000},
    {"employee_id": 103, "name": "Kavin", "department": "Finance", "salary": 50000},
]

try:
    with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=employees[0].keys())
        writer.writeheader()
        writer.writerows(employees)

    print("CSV file created successfully.")

    with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        print("\nEmployee Records:")
        for row in reader:
            print(row)

except FileNotFoundError:
    print("Error: File not found.")
except PermissionError:
    print("Error: Permission denied.")
except csv.Error as error:
    print("CSV error:", error)
except OSError as error:
    print("File error:", error)
```
