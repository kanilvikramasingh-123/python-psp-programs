# 📝 **Program No. 03 – TCP Client-Server Application with Echo**

## 🟢 **Creating the TCP Server:**
**The server creates a TCP socket, binds it to a local host and port, and listens for incoming client connections.**

## 🔵 **Connecting the Client:**
**The client creates its own TCP socket and connects to the server using the same host and port.**

## 🟡 **Sending the Message:**
**The client sends a message to the server after converting the text into bytes using UTF-8 encoding.**

## 🟠 **Echoing the Message:**
**The server receives the message and sends the same bytes back to the client, demonstrating the echo operation.**

## 🔴 **Closing the Connection:**
**Both sockets are closed after communication is completed, and exceptions are handled for common connection errors.**


```python
# Run this cell in one terminal/process as the server, then run the client
# by changing MODE to "client". Use only on your own/authorized lab host.

import socket

HOST = "127.0.0.1"
PORT = 5000
MESSAGE = "Hello from TCP client"
MODE = "server"  # Change to "client" for the client program

if MODE == "server":
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as server:
        server.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
        server.bind((HOST, PORT))
        server.listen(1)
        print(f"Server listening on {HOST}:{PORT}")

        conn, addr = server.accept()
        with conn:
            print("Client connected:", addr)
            data = conn.recv(1024)
            print("Received:", data.decode("utf-8"))
            conn.sendall(data)
            print("Echo sent.")

elif MODE == "client":
    try:
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as client:
            client.connect((HOST, PORT))
            client.sendall(MESSAGE.encode("utf-8"))
            response = client.recv(1024)
            print("Server echoed:", response.decode("utf-8"))
    except ConnectionRefusedError:
        print("Could not connect. Start the server first.")
```
