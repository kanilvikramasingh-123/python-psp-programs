# 📝 **Program No. 04 – Send an HTTP POST Request and Display Status Code and Headers**

## 🟢 **Preparing the POST Request:**
**The `urllib.request` module is used to construct an HTTP POST request without requiring an external package.**

## 🔵 **Encoding the Data:**
**The form data is encoded into bytes before it is sent to the permitted test endpoint.**

## 🟡 **Sending the Request:**
**The request is sent to an HTTP test service using `urlopen()`, and the server response is received.**

## 🟠 **Displaying Status and Headers:**
**The program displays the HTTP status code and response headers returned by the server.**

## 🔴 **Handling Request Errors:**
**Network and HTTP errors are handled with exceptions so that connection problems are reported clearly.**


```python
from urllib import request, parse
from urllib.error import HTTPError, URLError

# Public HTTP test endpoint. Replace with your permitted lab endpoint if needed.
URL = "https://httpbin.org/post"
DATA = {
    "name": "Student",
    "course": "Python Lab"
}

try:
    encoded_data = parse.urlencode(DATA).encode("utf-8")
    req = request.Request(URL, data=encoded_data, method="POST")

    with request.urlopen(req, timeout=10) as response:
        print("Status Code:", response.status)
        print("\nHeaders:")
        for key, value in response.headers.items():
            print(f"{key}: {value}")

except HTTPError as error:
    print("HTTP error:", error.code, error.reason)
except URLError as error:
    print("Connection error:", error.reason)
```
