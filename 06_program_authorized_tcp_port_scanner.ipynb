# 📝 **Program No. 06 – Scan Open TCP Ports on an Authorized Lab Host**

## 🟢 **Creating the Port Scanner:**
**The `socket` module is used to attempt TCP connections to ports on an authorized lab host.**

## 🔵 **Testing Individual Ports:**
**`connect_ex()` attempts a connection and returns a result code that indicates whether the connection succeeded.**

## 🟡 **Scanning a Port Range:**
**The program checks a defined range of TCP ports and records the ports that accept connections.**

## 🟠 **Identifying Services:**
**For open ports, `getservbyport()` is used when a standard TCP service name is available.**

## 🔴 **Displaying Authorized Lab Results:**
**The open ports and known service names are displayed. This program should only be used against systems and ranges you are authorized to test.**


```python
import socket

HOST = "127.0.0.1"   # Authorized lab host only
START_PORT = 1
END_PORT = 1024
TIMEOUT = 0.3

open_ports = []

for port in range(START_PORT, END_PORT + 1):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.settimeout(TIMEOUT)
        result = sock.connect_ex((HOST, port))

        if result == 0:
            try:
                service = socket.getservbyport(port, "tcp")
            except OSError:
                service = "unknown"
            open_ports.append((port, service))

print(f"Open TCP ports on {HOST}:")
if open_ports:
    for port, service in open_ports:
        print(f"Port {port}: {service}")
else:
    print("No open ports found in the selected range.")
```
