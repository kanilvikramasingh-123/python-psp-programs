# 📝 **Program No. 10 – Develop a Multithreaded TCP Server**

## 🟢 **Creating the TCP Server:**
**The `socket` module creates a TCP server that binds to a local host and port and listens for incoming clients.**

## 🔵 **Accepting Multiple Clients:**
**The server continuously accepts new connections. Each client is handled independently so one connection does not block the others.**

## 🟡 **Creating Client Threads:**
**The `threading` module creates a separate thread for every connected client, allowing simultaneous communication.**

## 🟠 **Handling Client Communication:**
**The client handler receives a message, displays it, and sends the same message back as an echo. The client can send `exit` to close its connection.**

## 🔴 **Handling Server Errors:**
**Network errors and keyboard interruption are handled so the server can stop safely.**


```python
import socket
import threading

HOST = "127.0.0.1"
PORT = 5000
BUFFER_SIZE = 1024

def handle_client(client_socket, client_address):
    print("Client connected:", client_address)

    with client_socket:
        while True:
            data = client_socket.recv(BUFFER_SIZE)

            if not data:
                break

            message = data.decode("utf-8")
            print(f"{client_address}: {message}")
            client_socket.sendall(data)

            if message.lower() == "exit":
                break

    print("Client disconnected:", client_address)

try:
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as server:
        server.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
        server.bind((HOST, PORT))
        server.listen()

        print(f"Server listening on {HOST}:{PORT}")

        while True:
            client_socket, client_address = server.accept()

            thread = threading.Thread(
                target=handle_client,
                args=(client_socket, client_address),
                daemon=True
            )
            thread.start()

            print("Active threads:", threading.active_count())

except OSError as error:
    print("Server error:", error)
except KeyboardInterrupt:
    print("\nServer stopped by the user.")
```
