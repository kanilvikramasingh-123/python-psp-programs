# 📝 **Program No. 16 – Extract Emails and IP Addresses from a Text File Using Regex**

## 🟢 **Creating Regular Expressions:**
**The `re` module is used to define patterns that can identify email addresses and IPv4 address-like strings in text.**

## 🔵 **Reading the Text File:**
**The program opens a text file and reads its complete content for pattern matching.**

## 🟡 **Extracting Email Addresses:**
**`re.findall()` searches the text using an email pattern and returns all matching email addresses.**

## 🟠 **Extracting IP Addresses:**
**A regular expression identifies IPv4 address formats, and the extracted values are collected for display.**

## 🔴 **Displaying Unique Results:**
**The program removes duplicates while preserving the discovered values and prints the email and IP address lists.**


```python
import re
from pathlib import Path

TEXT_FILE = Path("sample_text.txt")

TEXT_FILE.write_text(
    "Contact alice@example.com or bob@test.org.\n"
    "Lab server: 192.168.1.10 and DNS server: 8.8.8.8.\n"
    "Another address is admin@school.edu.\n",
    encoding="utf-8"
)

text = TEXT_FILE.read_text(encoding="utf-8")

email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
ip_pattern = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"

emails = list(dict.fromkeys(re.findall(email_pattern, text)))
ips = list(dict.fromkeys(re.findall(ip_pattern, text)))

print("Email addresses:")
for email in emails:
    print(email)

print("\nIP addresses:")
for ip in ips:
    print(ip)
```
