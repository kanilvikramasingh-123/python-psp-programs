# 📝 **Program No. 19 – Create a Network-Utility Module and Import It**

## 🟢 **Creating the Utility Module:**
**A separate Python module contains reusable network-related functions such as hostname resolution and local IP discovery.**

## 🔵 **Defining Reusable Functions:**
**Functions keep network operations organized and allow the same code to be reused by other programs.**

## 🟡 **Importing the Module:**
**The main program imports the custom module using Python's normal import mechanism.**

## 🟠 **Calling Utility Functions:**
**The imported functions are called from the main program to perform network operations without duplicating their implementation.**

## 🔴 **Displaying the Results:**
**The program prints the resolved address information and demonstrates how custom modules improve code organization and reuse.**


```python
# This notebook creates network_utility.py and then imports it.

from pathlib import Path

module_code = '''
import socket

def resolve_hostname(hostname):
    return socket.gethostbyname(hostname)

def local_hostname():
    return socket.gethostname()

def local_ip():
    return socket.gethostbyname(socket.gethostname())
'''

Path("network_utility.py").write_text(module_code, encoding="utf-8")

import network_utility

print("Local hostname:", network_utility.local_hostname())
print("Local IP:", network_utility.local_ip())

try:
    print("example.com IP:", network_utility.resolve_hostname("example.com"))
except socket.gaierror:
    print("Could not resolve example.com.")
```
