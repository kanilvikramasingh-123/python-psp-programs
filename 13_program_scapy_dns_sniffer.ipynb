# 📝 **Program No. 13 – Use Scapy to Sniff DNS Queries on the Lab Interface**

## 🟢 **Importing Scapy:**
**Scapy provides packet-capture and packet-inspection functions that can be used to observe network traffic in an authorized lab environment.**

## 🔵 **Selecting the Lab Interface:**
**The `INTERFACE` variable identifies the network interface used for the lab capture. It should be changed to the interface available in the lab machine.**

## 🟡 **Filtering DNS Traffic:**
**A BPF filter for UDP port 53 limits the capture to common DNS traffic instead of processing unrelated packets.**

## 🟠 **Extracting DNS Queries:**
**The callback checks for DNS query packets and reads the queried domain name from the `DNSQR` layer.**

## 🔴 **Displaying Captured Queries:**
**Each observed DNS query is printed. Packet sniffing must be performed only on interfaces and networks where you have authorization.**


```python
# Install Scapy if needed:
# %pip install scapy

from scapy.all import sniff, DNS, DNSQR

INTERFACE = "eth0"  # Change to your authorized lab interface

def process_packet(packet):
    if packet.haslayer(DNS) and packet.haslayer(DNSQR):
        if packet[DNS].qr == 0:  # 0 means query
            query = packet[DNSQR].qname.decode("utf-8", errors="ignore")
            print("DNS Query:", query)

print(f"Sniffing DNS queries on {INTERFACE}...")
print("Use Ctrl+C to stop.")

try:
    sniff(
        iface=INTERFACE,
        filter="udp port 53",
        prn=process_packet,
        store=False
    )
except PermissionError:
    print("Permission denied. Run the notebook with appropriate capture privileges.")
except OSError as error:
    print("Capture error:", error)
```
