# 📝 **Program No. 05 – Resolve a Domain Name and Perform a Reverse Lookup**

## 🟢 **Importing the Socket Module:**
**Python's `socket` module provides functions for network name resolution and reverse DNS lookups.**

## 🔵 **Forward DNS Lookup:**
**`socket.gethostbyname()` converts a domain name such as `example.com` into an IPv4 address.**

## 🟡 **Reverse DNS Lookup:**
**`socket.gethostbyaddr()` performs a reverse lookup by taking an IP address and returning the associated hostname information.**

## 🟠 **Displaying the Results:**
**The resolved IP address and reverse lookup hostname are printed clearly for comparison.**

## 🔴 **Handling DNS Errors:**
**`socket.gaierror` and `socket.herror` are handled to report invalid domains or unavailable DNS records.**


```python
import socket

domain = "example.com"

try:
    ip_address = socket.gethostbyname(domain)
    print("Domain:", domain)
    print("IP Address:", ip_address)

    host_name, aliases, addresses = socket.gethostbyaddr(ip_address)
    print("\nReverse Lookup:")
    print("Hostname:", host_name)
    print("Aliases:", aliases)
    print("Addresses:", addresses)

except socket.gaierror as error:
    print("DNS lookup error:", error)
except socket.herror as error:
    print("Reverse lookup error:", error)
```
