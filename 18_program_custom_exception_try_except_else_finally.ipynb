# 📝 **Program No. 18 – Demonstrate a Custom Exception with try-except-else-finally**

## 🟢 **Creating a Custom Exception:**
**A class named `InvalidAgeError` is created by inheriting from Python's `Exception` class. This provides a meaningful application-specific error.**

## 🔵 **Raising the Exception:**
**The program raises the custom exception when the supplied age does not satisfy the required condition.**

## 🟡 **Using try-except:**
**The `try` block contains the operation that may fail, while the `except` block catches and displays the custom exception.**

## 🟠 **Using else:**
**The `else` block executes only when no exception occurs, confirming that the operation completed successfully.**

## 🔴 **Using finally:**
**The `finally` block executes whether an exception occurs or not, making it suitable for cleanup or final status messages.**


```python
class InvalidAgeError(Exception):
    pass

age = 20

try:
    if age < 18:
        raise InvalidAgeError("Age must be 18 or above.")
except InvalidAgeError as error:
    print("Custom exception:", error)
else:
    print("No exception: age is valid.")
finally:
    print("Finally block executed.")
```
