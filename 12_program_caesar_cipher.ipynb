# 📝 **Program No. 12 – Implement Caesar Cipher for Encryption and Decryption**

## 🟢 **Defining the Caesar Cipher Function:**
**The `caesar_cipher()` function shifts alphabetic characters by a specified number of positions while preserving uppercase and lowercase letters.**

## 🔵 **Setting the Message and Shift:**
**The program directly assigns the message and shift value, so no user input is required.**

## 🟡 **Performing Encryption:**
**During encryption, each letter is shifted forward by the selected value. For example, with a shift of 3, `Hello` becomes `Khoor`.**

## 🟠 **Performing Decryption:**
**Decryption uses the negative shift value to move characters back to their original positions.**

## 🔴 **Displaying the Results:**
**The original, encrypted, and decrypted messages are printed to demonstrate successful encryption and recovery.**


```python
def caesar_cipher(text, shift):
    result = ""

    for char in text:
        if char.isalpha():
            if char.isupper():
                result += chr((ord(char) - ord("A") + shift) % 26 + ord("A"))
            else:
                result += chr((ord(char) - ord("a") + shift) % 26 + ord("a"))
        else:
            result += char

    return result

message = "Hello World"
shift = 3

encrypted_text = caesar_cipher(message, shift)
decrypted_text = caesar_cipher(encrypted_text, -shift)

print("Original text:", message)
print("Shift value:", shift)
print("Encrypted text:", encrypted_text)
print("Decrypted text:", decrypted_text)
```
