# 📝 **Program No. 20 – Demonstrate Secure Coding: Input Sanitization, Whitelisting, and Avoiding eval()**

## 🟢 **Input Sanitization:**
**The program strips unnecessary whitespace and validates the input before using it, reducing unexpected or unsafe input handling.**

## 🔵 **Using a Whitelist:**
**Only values present in an explicit allow-list are accepted. This is safer than accepting arbitrary values.**

## 🟡 **Validating Numeric Input:**
**The program validates numeric text before converting it to an integer, preventing invalid values from causing unexpected errors.**

## 🟠 **Avoiding eval():**
**The program does not execute user-provided strings as Python code. `eval()` can execute arbitrary expressions and should not be used for untrusted input.**

## 🔴 **Applying Safe Alternatives:**
**For fixed choices, whitelists and explicit parsing are preferred. For structured data, safe parsers such as `json.loads()` should be used instead of executing input as code.**


```python
import json

# Example input values. In real applications these could come from a form/API.
username = " student_01 "
role = "student"
age_text = "20"
json_text = '{"course": "Python", "level": "beginner"}'

# 1. Input sanitization
username = username.strip()

# 2. Whitelisting
allowed_roles = {"student", "teacher", "admin"}
if role not in allowed_roles:
    raise ValueError("Invalid role.")

# 3. Safe numeric validation
if not age_text.isdigit():
    raise ValueError("Age must contain digits.")

age = int(age_text)
if not 1 <= age <= 120:
    raise ValueError("Age is outside the allowed range.")

# 4. Safe parsing instead of eval()
data = json.loads(json_text)

print("Sanitized username:", username)
print("Whitelisted role:", role)
print("Validated age:", age)
print("Safely parsed JSON:", data)

print("\nSecurity rule: Never use eval() on untrusted user input.")
```
