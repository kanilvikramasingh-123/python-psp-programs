# 📝 **Program No. 17 – Use Raw Sockets to Fetch HTTP Headers from a Permitted Lab Server**

## 🟢 **Creating a Raw TCP Socket:**
**The program creates a normal TCP socket and connects to a permitted lab HTTP server on port 80. This demonstrates HTTP at the socket level.**

## 🔵 **Building the HTTP Request:**
**A minimal HTTP `HEAD` request is constructed manually with the required `Host` and `Connection` headers.**

## 🟡 **Sending the Request:**
**The request is encoded into bytes and transmitted through the established TCP connection.**

## 🟠 **Receiving HTTP Headers:**
**The server response is received and displayed. A `HEAD` request asks for response headers without requesting the full response body.**

## 🔴 **Handling Network Errors:**
**Socket timeouts and connection errors are handled. Use this only with a server you own or are explicitly permitted to test.**


```python
import socket

HOST = "example.com"  # Replace with your permitted lab server
PORT = 80
TIMEOUT = 5

request = (
    f"HEAD / HTTP/1.1\r\n"
    f"Host: {HOST}\r\n"
    f"Connection: close\r\n"
    f"\r\n"
)

try:
    with socket.create_connection((HOST, PORT), timeout=TIMEOUT) as sock:
        sock.sendall(request.encode("ascii"))

        response = b""
        while True:
            chunk = sock.recv(4096)
            if not chunk:
                break
            response += chunk

    headers = response.decode("iso-8859-1", errors="replace")
    header_section = headers.split("\r\n\r\n", 1)[0]

    print("HTTP response headers:")
    print(header_section)

except socket.timeout:
    print("Connection timed out.")
except OSError as error:
    print("Socket error:", error)
```
