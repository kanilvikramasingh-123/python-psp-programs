# 📝 **Program No. 15 – Parse a Web Server Access Log and Report the Top 5 IPs**

## 🟢 **Reading the Access Log:**
**The program reads a web server access log line by line and extracts the client IP address from each valid entry.**

## 🔵 **Counting IP Addresses:**
**A dictionary or `Counter` is used to count how many requests were made by each IP address.**

## 🟡 **Parsing Log Entries:**
**The first whitespace-separated field in a common access-log format is treated as the client IP address.**

## 🟠 **Finding the Top 5 IPs:**
**The counts are sorted in descending order so the five IP addresses with the most requests can be identified.**

## 🔴 **Displaying the Report:**
**The program prints the top five IP addresses and their request counts, while safely skipping malformed lines.**


```python
from collections import Counter
from pathlib import Path

LOG_FILE = Path("access.log")

# Sample access log for demonstration.
LOG_FILE.write_text(
    "192.168.1.10 - - [08/Oct/2026:10:00:00] \"GET / HTTP/1.1\" 200 1024\n"
    "192.168.1.20 - - [08/Oct/2026:10:01:00] \"GET /login HTTP/1.1\" 200 512\n"
    "192.168.1.10 - - [08/Oct/2026:10:02:00] \"GET /home HTTP/1.1\" 200 900\n"
    "10.0.0.5 - - [08/Oct/2026:10:03:00] \"GET /api HTTP/1.1\" 200 1200\n"
    "192.168.1.10 - - [08/Oct/2026:10:04:00] \"GET /about HTTP/1.1\" 200 700\n"
    "192.168.1.20 - - [08/Oct/2026:10:05:00] \"GET /contact HTTP/1.1\" 200 600\n"
    "10.0.0.5 - - [08/Oct/2026:10:06:00] \"GET /docs HTTP/1.1\" 200 800\n",
    encoding="utf-8"
)

counts = Counter()

with LOG_FILE.open("r", encoding="utf-8") as file:
    for line in file:
        parts = line.split()
        if parts:
            counts[parts[0]] += 1

print("Top 5 IP addresses:")
for ip, count in counts.most_common(5):
    print(f"{ip}: {count} requests")
```
