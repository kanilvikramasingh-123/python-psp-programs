# 📝 **Program No. 02 – Store Product Details in a Dictionary of Lists and Display Them in a Table**

## 🟢 **Creating Product Data:**
**Product details are stored in a dictionary where each key represents a column and each value is a list of product records.**

## 🔵 **Getting Column Information:**
**The program obtains the column headings and determines the number of product records stored in the lists.**

## 🟡 **Calculating Column Widths:**
**The width of each column is calculated from the longest value so that the table is aligned neatly.**

## 🟠 **Creating the Table:**
**A separator, headings, and product rows are printed using formatted strings to create a structured table.**

## 🔴 **Displaying Product Records:**
**All product details, including ID, name, category, price, and stock, are displayed in an easy-to-read tabular format.**


```python
products = {
    "Product ID": [101, 102, 103, 104],
    "Product Name": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics"],
    "Price": [55000, 800, 1500, 12000],
    "Stock": [10, 50, 30, 15]
}

headers = list(products.keys())
rows = list(zip(*products.values()))

widths = {
    header: max(len(str(header)), *(len(str(value)) for value in products[header]))
    for header in headers
}

separator = "+".join("-" * (widths[h] + 2) for h in headers)
separator = "+" + separator + "+"

print(separator)
print("| " + " | ".join(str(h).ljust(widths[h]) for h in headers) + " |")
print(separator)

for row in rows:
    print("| " + " | ".join(str(value).ljust(widths[h]) for h, value in zip(headers, row)) + " |")

print(separator)
```
