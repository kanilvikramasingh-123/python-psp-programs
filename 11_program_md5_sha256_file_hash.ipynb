# 📝 **Program No. 11 – Generate SHA-256 and MD5 Hashes of a File to Verify Integrity**

## 🟢 **Importing the Hashing Module:**
**The `hashlib` module is imported to generate cryptographic hash values using MD5 and SHA-256 algorithms.**

## 🔵 **Opening and Reading the File:**
**The file is opened in binary mode and read in small chunks, which allows large files to be processed efficiently.**

## 🟡 **Generating MD5 and SHA-256 Hashes:**
**Each file chunk is supplied to both hash objects, and `hexdigest()` converts the final digest into readable hexadecimal text.**

## 🟠 **Displaying the Hash Values:**
**The program displays the file name and its MD5 and SHA-256 checksums for later comparison.**

## 🔴 **Verifying File Integrity:**
**If a calculated hash matches a trusted previously recorded hash, the file contents are considered unchanged. A different hash indicates that the contents have changed or become corrupted.**


```python
import hashlib
from pathlib import Path

FILE_NAME = "example.txt"

# Create a sample file for demonstration.
Path(FILE_NAME).write_text("This is a sample file for integrity checking.\n", encoding="utf-8")

md5_hash = hashlib.md5()
sha256_hash = hashlib.sha256()

try:
    with open(FILE_NAME, "rb") as file:
        while chunk := file.read(4096):
            md5_hash.update(chunk)
            sha256_hash.update(chunk)

    print("File:", FILE_NAME)
    print("MD5 Hash:", md5_hash.hexdigest())
    print("SHA-256 Hash:", sha256_hash.hexdigest())

except FileNotFoundError:
    print("Error: File not found.")
except OSError as error:
    print("File error:", error)
```
