# 📝 **Program No. 14 – Check Password Strength and Display a Rating**

## 🟢 **Checking Password Length:**
**The program first checks the password length because longer passwords generally provide a larger search space than very short passwords.**

## 🔵 **Checking Character Categories:**
**The password is tested for uppercase letters, lowercase letters, digits, and special characters to measure its composition.**

## 🟡 **Calculating the Strength Score:**
**Points are assigned for length and character diversity, and the total score is used to determine the strength rating.**

## 🟠 **Displaying the Rating:**
**The program reports a simple rating such as Very Weak, Weak, Moderate, Strong, or Very Strong.**

## 🔴 **Providing Security Guidance:**
**The program reminds users to use long, unique passwords and avoid personal information or reused passwords.**


```python
import string

password = "Lab@Python2026"

score = 0

# Length score
if len(password) >= 8:
    score += 1
if len(password) >= 12:
    score += 1
if len(password) >= 16:
    score += 1

# Character diversity
if any(c.islower() for c in password):
    score += 1
if any(c.isupper() for c in password):
    score += 1
if any(c.isdigit() for c in password):
    score += 1
if any(c in string.punctuation for c in password):
    score += 1

if score <= 2:
    rating = "Very Weak"
elif score <= 4:
    rating = "Weak"
elif score <= 5:
    rating = "Moderate"
elif score <= 6:
    rating = "Strong"
else:
    rating = "Very Strong"

print("Password length:", len(password))
print("Strength score:", score, "/ 7")
print("Password rating:", rating)
print("\nTip: Use a long, unique password and avoid personal information.")
```
