# 📝 **Program No. 08 – Count ERROR and WARNING Lines in a Log File**

## 🟢 **Opening the Log File:**
**The program opens `log.txt` in read mode and processes the file line by line.**

## 🔵 **Reading Log Entries:**
**Each line is examined individually so the program can identify messages containing the required keywords.**

## 🟡 **Counting ERROR Lines:**
**Whenever the word `ERROR` occurs in a line, the error counter is increased by one.**

## 🟠 **Counting WARNING Lines:**
**Whenever the word `WARNING` occurs in a line, the warning counter is increased by one.**

## 🔴 **Displaying Results and Handling Errors:**
**The final counts are displayed, and `FileNotFoundError` is handled if the log file is missing.**


```python
from pathlib import Path

LOG_FILE = Path("log.txt")

# Create a sample log file for demonstration.
LOG_FILE.write_text(
    "INFO Server started\n"
    "ERROR Database connection failed\n"
    "WARNING Low disk space\n"
    "ERROR Authentication failed\n"
    "INFO Request completed\n"
    "WARNING High memory usage\n",
    encoding="utf-8"
)

error_count = 0
warning_count = 0

try:
    with LOG_FILE.open("r", encoding="utf-8") as file:
        for line in file:
            if "ERROR" in line:
                error_count += 1
            if "WARNING" in line:
                warning_count += 1

    print("Number of ERROR lines:", error_count)
    print("Number of WARNING lines:", warning_count)

except FileNotFoundError:
    print("Error: log.txt file not found.")
```
